## 6. Karşılaştırma, mantık ve karar verme — 12 dk

Karşılaştırmaların sonucu `True` veya `False` olur.

| Kod | Anlamı |
|---|---|
| `a == b` | eşit mi? |
| `a != b` | farklı mı? |
| `a > b` / `a < b` | büyük mü / küçük mü? |
| `a >= b` / `a <= b` | büyük-eşit / küçük-eşit |
| `and` / `or` | ve / veya |

**`=` atama yapar, `==` karşılaştırır.**

```python
puan = 72
print(puan >= 50)
print(puan == 100)
print(puan >= 50 and puan <= 100)
print(puan < 0 or puan > 100)
```

`if` koşul doğruysa belirli kodu çalıştırır; `elif` diğer koşulları, `else` geriye kalan durumları ele alır.

**Girintiye dikkat!** `if` altındaki satırların başında aynı miktarda boşluk (genelde 4) olmalıdır.

```python
notu = 72
if notu >= 85:
    print("Pekiyi")
elif notu >= 50:
    print("Geçti")
else:
    print("Tekrar deneyelim")
```

## 7. Döngüler: aynı işi tekrar yapma — 12 dk

`for` bir değer dizisi üzerinde ilerler. `range(5)` bize **0, 1, 2, 3, 4** değerlerini sağlar; son sınır dahil değildir.

`range(1, 6)` → **1, 2, 3, 4, 5**. Döngü gövdesinde yine girinti kullanılır.

```python
for sayi in range(5):
    print("Sayı:", sayi)

print("---")
for sayi in range(1, 6):
    print(sayi)
```

Biriktirme örneği: toplama sıfırdan başlar; her adımda yeni sayı eklenir.

```python
toplam = 0
for sayi in range(1, 6):
    toplam += sayi
print("1'den 5'e toplam:", toplam)
```

`while` bir koşul **doğru kaldığı sürece** döner. Döngü içinde koşulun değişmesini sağlamazsan sonsuz döngü oluşabilir.

```python
sayac = 3
while sayac > 0:
    print("Geri sayım:", sayac)
    sayac -= 1
print("Başla!")
```

### ✍️ Sen dene: Çift sayıları yazdır

`range(0, 11)` ve `if` kullanarak **0 ile 10 arasındaki çift sayıları** yazdır. Bir sayının 2 ile bölümünden kalanını `%` ile bulabilirsin.

**Beklenen sonuç:** 0, 2, 4, 6, 8, 10 (ayrı satırlarda).

<details><summary>İpucu (takılırsan aç)</summary>

Çift sayı için `sayi % 2 == 0`.

</details>

```python
for sayi in range(0, 11):
    if sayi % 2 == 0:
        print(sayi)
```

## 8. Listeler — 9 dk

Liste, birden fazla değeri tek isim altında tutar. Köşeli parantezle oluşturulur. **İndeksler 0'dan başlar.**

- `liste[0]`: ilk eleman
- `liste[-1]`: son eleman
- `len(liste)`: kaç eleman var?
- `liste.append(deger)`: sona eleman ekler

```python
notlar = [70, 80, 90]
print("İlk not:", notlar[0])
print("Son not:", notlar[-1])
print("Not sayısı:", len(notlar))

notlar.append(100)
print("Yeni liste:", notlar)

for notu in notlar:
    print("Listedeki not:", notu)
```

### ✍️ Sen dene: Liste ortalaması

`notlar = [60, 80, 100]` listesindeki sayıları **for döngüsüyle topla**. Eleman sayısını `len()` ile öğrenip ortalamayı hesapla.

**Beklenen sonuç:** Ortalama: 80.0

```python
notlar = [60, 80, 100]
toplam = 0
for notu in notlar:
    toplam += notu
ortalama = toplam / len(notlar)
print("Ortalama:", ortalama)
```

> **Hoca notu / çözüm:** Bu alıştırmanın çalışan çözümü yalnızca öğretmen dosyasındaki bir önceki kod hücresindedir.

## 9. 🎯 Mini proje: Basit market sepeti — 12 dk

Artık birden çok konuyu **tek bir küçük programda** birleştireceğiz.

**Senaryo:** Sepette üç ürün var. Ürün adları ve fiyatları iki listede, aynı sırayla tutuluyor. Sepet **100 TL veya üzeriyse %10 indirim** uygulanıyor. Müşterinin bütçesi **100 TL**.

Programın şu işleri yapmalı:

1. Her ürünü ve fiyatını göster.
2. Fiyatları bir `for` döngüsüyle topla.
3. Toplam 100 TL veya daha yüksekse %10 indirim hesapla; değilse indirim 0 olsun.
4. Ödenecek tutarı hesapla.
5. Bütçe yetiyorsa kalan parayı; yetmiyorsa eksik tutarı yazdır.

**Başlangıç verileri:** Defter 60 TL, Kalem 35 TL, Silgi 15 TL → toplam **110 TL**; indirim **11 TL**; ödenecek **99 TL**; kalan **1 TL**.

> **İpucu:** Aynı konumdaki ad ve fiyat için `for i in range(len(urunler)):` kullanıp `urunler[i]` ve `fiyatlar[i]` okuyabilirsin. Önce sadece 1. ve 2. adımı çalıştır; sonra indirim ve bütçe kontrolünü ekle.

### ✍️ Sen dene: Market sepetini tamamla

Aşağıdaki verileri bozmadan eksik adımları kodla. Ürünleri tek tek yazdıktan sonra toplam, indirim, ödenecek tutar ve bütçeyi değerlendir.

**Beklenen sonuç:** Toplam 110.0 değil, 110 TL; indirim 11.0 TL; ödenecek 99.0 TL; kalan 1.0 TL.

<details><summary>İpucu (takılırsan aç)</summary>

Aynı anda üç konu gerekiyor: `for`, `if`, listeden `fiyatlar[i]` erişimi.

</details>

```python
urunler = ["Defter", "Kalem", "Silgi"]
fiyatlar = [60, 35, 15]
butce = 100

toplam = 0
for i in range(len(urunler)):
    print(f"{urunler[i]}: {fiyatlar[i]} TL")
    toplam += fiyatlar[i]

if toplam >= 100:
    indirim = toplam * 0.10
else:
    indirim = 0

odenecek = toplam - indirim
print(f"Toplam: {toplam} TL")
print(f"İndirim: {indirim} TL")
print(f"Ödenecek: {odenecek} TL")

if butce >= odenecek:
    print(f"Bütçe yeterli. Kalan: {butce - odenecek} TL")
else:
    print(f"Bütçe yetersiz. Eksik: {odenecek - butce} TL")
```

> **Hoca notu / çözüm:** Bu alıştırmanın çalışan çözümü yalnızca öğretmen dosyasındaki bir önceki kod hücresindedir.
