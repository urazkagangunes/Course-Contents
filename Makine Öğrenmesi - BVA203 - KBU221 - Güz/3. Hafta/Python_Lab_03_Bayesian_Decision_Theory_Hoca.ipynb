# 1. Olasılık Tekrarı

\[

\]

Örnek:

Bir sınıfta 40 öğrenci var ve 10 öğrenci gözlüklü.

\[
P(Gözlüklü) = \frac{10}{40} = 0.25

Yani rastgele seçilen bir öğrencinin gözlüklü olma olasılığı **%25**.

````python
toplam = 40
gozluklu = 10

p_gozluklu = gozluklu / toplam

print("P(Gözlüklü) =", p_gozluklu)
````

# 3. Prior — Önsel Olasılık

- 100 e-posta

ise:

\[

\]

Bunlar **prior probability** değerleridir.

````python
spam = 30
normal = 70
toplam = spam + normal

p_spam = spam / toplam
p_normal = normal / toplam

print("P(Spam) =", p_spam)
print("P(Normal) =", p_normal)
````

# 4. Likelihood — Olabilirlik

Şimdi yeni e-postada:

"kazandınız"

Daha önceki spam e-postaların %80'inde bu kelime geçiyor olsun:

````python
p_kelime_given_spam = 0.80
p_kelime_given_normal = 0.10

print("P(Kelime | Spam) =", p_kelime_given_spam)
print("P(Kelime | Normal) =", p_kelime_given_normal)
````

# 5. Bayes Teoremi

Bayes teoremi:

\[
P(C|x) = \frac{P(x|C)P(C)}{P(x)}
\]

Burada:

- \(P(C)\) → Prior
- \(P(x|C)\) → Likelihood
- \(P(C|x)\) → Posterior
- \(P(x)\) → Evidence

Bizim asıl istediğimiz şey **posterior**:

> Gözlemi gördükten sonra sınıfın olasılığı nedir?

## Spam örneği

Elimizde:

\[
P(Spam)=0.30
\]

\[
P(Normal)=0.70
\]

\[
P(Kelime|Spam)=0.80
\]

\[
P(Kelime|Normal)=0.10
\]

Önce kanıtın toplam olasılığını hesaplarız:

\[
P(Kelime)=P(Kelime|Spam)P(Spam)+P(Kelime|Normal)P(Normal)
\]

````python
p_spam = 0.30
p_normal = 0.70

p_kelime_given_spam = 0.80
p_kelime_given_normal = 0.10

p_kelime = (
    p_kelime_given_spam * p_spam
    +
    p_kelime_given_normal * p_normal
)

print("P(Kelime) =", p_kelime)
````

Şimdi posterior:

\[
P(Spam|Kelime)
=
\frac{P(Kelime|Spam)P(Spam)}
{P(Kelime)}
\]

````python
p_spam_given_kelime = (
    p_kelime_given_spam * p_spam
) / p_kelime

p_normal_given_kelime = (
    p_kelime_given_normal * p_normal
) / p_kelime

print("P(Spam | Kelime) =", p_spam_given_kelime)
print("P(Normal | Kelime) =", p_normal_given_kelime)
````

### Beklenen yorum

Yeni kelimeyi görmeden önce spam olasılığı:

```text
0.30
```

Kelimeyi gördükten sonra spam olasılığı yaklaşık:

```text
0.774
```

oldu.

Yani yeni kanıt, kararımızı ciddi biçimde değiştirdi.

> **Bayes'in temel mantığı:** Yeni bilgi geldikçe inancımızı güncelleriz.

# 6. Bayes Hesabını Fonksiyonlaştırma

Aynı hesabı tekrar tekrar yazmamak için fonksiyon oluşturalım.

````python
def posterior_hesapla(prior_c1, prior_c2, likelihood_c1, likelihood_c2):

    evidence = (
        likelihood_c1 * prior_c1
        +
        likelihood_c2 * prior_c2
    )

    posterior_c1 = (likelihood_c1 * prior_c1) / evidence
    posterior_c2 = (likelihood_c2 * prior_c2) / evidence

    return posterior_c1, posterior_c2


spam_post, normal_post = posterior_hesapla(
    0.30,
    0.70,
    0.80,
    0.10
)

print("Spam posterior:", spam_post)
print("Normal posterior:", normal_post)
````

# 7. MAP Karar Kuralı

**MAP = Maximum A Posteriori**

En yüksek posterior olasılığa sahip sınıf seçilir.

Yani:

\[
P(C_1|x) > P(C_2|x)
\]

ise \(C_1\) sınıfını seçeriz.

Spam örneğinde:

```text
P(Spam | kelime) ≈ 0.774
P(Normal | kelime) ≈ 0.226
```

Bu nedenle karar:

```text
Spam
```

````python
spam_posterior = 0.774
normal_posterior = 0.226

if spam_posterior > normal_posterior:
    print("Karar: SPAM")
else:
    print("Karar: NORMAL")
````

# 8. Posterior İçin Paydadan Kurtulabilir miyiz?

İki sınıfı karşılaştırırken:

\[
P(C|x) = \frac{P(x|C)P(C)}{P(x)}
\]

ifadesindeki \(P(x)\) her iki sınıf için de aynıdır.

Bu nedenle yalnızca şu değerleri karşılaştırabiliriz:

\[
P(x|C_1)P(C_1)
\]

ve

\[
P(x|C_2)P(C_2)
\]

Bu hesap daha pratiktir.

````python
p_spam = 0.30
p_normal = 0.70

likelihood_spam = 0.80
likelihood_normal = 0.10

spam_skor = likelihood_spam * p_spam
normal_skor = likelihood_normal * p_normal

print("Spam skoru:", spam_skor)
print("Normal skoru:", normal_skor)

if spam_skor > normal_skor:
    print("MAP kararı: Spam")
else:
    print("MAP kararı: Normal")
````

# 9. Birden Fazla Özellik ile Karar Verme

Bir e-posta için iki özellik düşünelim:

- "kazandınız" kelimesi var mı?
- "ücretsiz" kelimesi var mı?

Basitleştirmek için özelliklerin sınıf verildiğinde bağımsız olduğunu varsayalım.

O zaman:

\[
P(x_1,x_2|C)
=
P(x_1|C)P(x_2|C)
\]

Bu fikir daha sonra **Naive Bayes** yönteminde karşımıza çıkacaktır.

````python
# Priorlar
p_spam = 0.30
p_normal = 0.70

# Özellik 1: "kazandınız"
p_kazandiniz_given_spam = 0.80
p_kazandiniz_given_normal = 0.10

# Özellik 2: "ücretsiz"
p_ucretsiz_given_spam = 0.70
p_ucretsiz_given_normal = 0.20

spam_skor = (
    p_spam
    * p_kazandiniz_given_spam
    * p_ucretsiz_given_spam
)

normal_skor = (
    p_normal
    * p_kazandiniz_given_normal
    * p_ucretsiz_given_normal
)

print("Spam skoru:", spam_skor)
print("Normal skoru:", normal_skor)

if spam_skor > normal_skor:
    print("Karar: Spam")
else:
    print("Karar: Normal")
````

# 12. Expected Risk — Beklenen Risk

Bir karar vermeden önce olası hataların maliyetini posterior olasılıklarla birlikte düşünürüz.

İki kararımız olsun:

- Karar 1: Hasta de
- Karar 2: Sağlıklı de

Posterior değerler:

\[
P(Hasta|x)=0.30
\]

\[
P(Sağlıklı|x)=0.70
\]

İlk bakışta MAP:

```text
Sağlıklı
```

der.

Ama maliyetleri hesaba katarsak karar değişebilir.

````python
p_hasta = 0.30
p_saglikli = 0.70

# Kayıplar
# Hasta dediğimizde sağlıklıysa maliyet = 1
# Sağlıklı dediğimizde hastaysa maliyet = 10

risk_hasta_de = 1 * p_saglikli
risk_saglikli_de = 10 * p_hasta

print("Hasta de kararının riski:", risk_hasta_de)
print("Sağlıklı de kararının riski:", risk_saglikli_de)

if risk_hasta_de < risk_saglikli_de:
    print("Minimum risk kararı: HASTA")
else:
    print("Minimum risk kararı: SAĞLIKLI")
````

# 14. MAP ve Minimum Risk Farkı

## MAP

En yüksek posterioru seçer:

\[
\arg\max_C P(C|x)
\]

## Minimum Risk

Beklenen kaybı en düşük olan kararı seçer:

\[
\arg\min_{\alpha_i} R(\alpha_i|x)
\]

Eğer bütün yanlış kararların maliyeti aynıysa, minimum risk kararı MAP ile aynı sonuca gelir.

Ama maliyetler farklıysa sonuç değişebilir.

# 15. Mini Uygulama — Meyve Sınıflandırma

İki sınıfımız olsun:

```text
Elma
Portakal
```

Bir meyvenin ağırlığını gözlemliyoruz.

Bu laboratuvar için likelihood değerlerini hazır vereceğiz.

Yeni meyve için:

```text
P(x | Elma) = 0.40
P(x | Portakal) = 0.20
```

Priorlar:

```text
P(Elma) = 0.60
P(Portakal) = 0.40
```

````python
p_elma = 0.60
p_portakal = 0.40

likelihood_elma = 0.40
likelihood_portakal = 0.20

elma_skor = p_elma * likelihood_elma
portakal_skor = p_portakal * likelihood_portakal

print("Elma skoru:", elma_skor)
print("Portakal skoru:", portakal_skor)

if elma_skor > portakal_skor:
    print("Tahmin: Elma")
else:
    print("Tahmin: Portakal")
````

# 16. Evidence Hesaplayıp Gerçek Posteriorları Bulalım

````python
evidence = elma_skor + portakal_skor

posterior_elma = elma_skor / evidence
posterior_portakal = portakal_skor / evidence

print("P(Elma | x) =", posterior_elma)
print("P(Portakal | x) =", posterior_portakal)
print("Toplam =", posterior_elma + posterior_portakal)
````

# 17. Küçük Veri Kümesinden Prior Hesaplama

````python
etiketler = [
    "Elma", "Elma", "Portakal",
    "Elma", "Portakal", "Elma",
    "Elma", "Portakal", "Elma", "Elma"
]

elma_sayisi = 0
portakal_sayisi = 0

for etiket in etiketler:
    if etiket == "Elma":
        elma_sayisi += 1
    else:
        portakal_sayisi += 1

toplam = len(etiketler)

p_elma = elma_sayisi / toplam
p_portakal = portakal_sayisi / toplam

print("Elma sayısı:", elma_sayisi)
print("Portakal sayısı:", portakal_sayisi)
print("P(Elma):", p_elma)
print("P(Portakal):", p_portakal)
````

# 19. Mini Proje — Basit Bayes Sınıflandırıcı

Şimdi küçük bir e-posta sınıflandırıcı yapacağız.

İki sınıf:

```text
Spam
Normal
```

Yeni e-postada iki özellik gözlemliyoruz:

```text
"ücretsiz" var
"kazandınız" var
```

Priorlar ve likelihood değerleri geçmiş veriden elde edilmiş olsun.

````python
# Priorlar

p_spam = 0.25
p_normal = 0.75

# "ücretsiz" özelliği

p_ucretsiz_spam = 0.80
p_ucretsiz_normal = 0.10

# "kazandınız" özelliği

p_kazandiniz_spam = 0.70
p_kazandiniz_normal = 0.05
````

````python
# Sınıf skorlarını hesaplayalım

spam_skor = (
    p_spam
    * p_ucretsiz_spam
    * p_kazandiniz_spam
)

normal_skor = (
    p_normal
    * p_ucretsiz_normal
    * p_kazandiniz_normal
)

print("Spam skoru:", spam_skor)
print("Normal skoru:", normal_skor)
````

````python
# Posteriorları normalize edelim

evidence = spam_skor + normal_skor

posterior_spam = spam_skor / evidence
posterior_normal = normal_skor / evidence

print("P(Spam | x):", posterior_spam)
print("P(Normal | x):", posterior_normal)
````

````python
# MAP kararı

if posterior_spam > posterior_normal:
    karar = "Spam"
else:
    karar = "Normal"

print("MAP kararı:", karar)
````

# 20. Aynı Mini Projede Maliyet Ekleyelim

Şimdi varsayalım:

- Normal e-postayı spam klasörüne atmanın maliyeti = 5
- Spam e-postayı gelen kutusuna bırakmanın maliyeti = 1

Bu durumda kararımız posteriora göre değil, minimum riske göre değişebilir.

````python
# Kararlar:
# Spam de
# Normal de

loss_normal_maili_spam_yap = 5
loss_spami_normal_yap = 1

# Spam dersek hata yalnızca gerçek sınıf Normal olduğunda oluşur
risk_spam_de = (
    loss_normal_maili_spam_yap
    * posterior_normal
)

# Normal dersek hata yalnızca gerçek sınıf Spam olduğunda oluşur
risk_normal_de = (
    loss_spami_normal_yap
    * posterior_spam
)

print("Spam de risk:", risk_spam_de)
print("Normal de risk:", risk_normal_de)

if risk_spam_de < risk_normal_de:
    print("Minimum risk kararı: Spam")
else:
    print("Minimum risk kararı: Normal")
````

# 21. Sınıf İçi Mini Alıştırma

Aşağıdaki değerleri kullanın:

```text
P(C1) = 0.40
P(C2) = 0.60

P(x | C1) = 0.70
P(x | C2) = 0.20
```

Görevler:

1. C1 skorunu bulun.
2. C2 skorunu bulun.
3. Evidence hesaplayın.
4. Posteriorları bulun.
5. MAP kararını verin.

````python
# Hoca çözümü

p_c1 = 0.40
p_c2 = 0.60

likelihood_c1 = 0.70
likelihood_c2 = 0.20

score_c1 = p_c1 * likelihood_c1
score_c2 = p_c2 * likelihood_c2

evidence = score_c1 + score_c2

posterior_c1 = score_c1 / evidence
posterior_c2 = score_c2 / evidence

print("C1 skoru:", score_c1)
print("C2 skoru:", score_c2)
print("Evidence:", evidence)
print("P(C1 | x):", posterior_c1)
print("P(C2 | x):", posterior_c2)

if posterior_c1 > posterior_c2:
    print("MAP kararı: C1")
else:
    print("MAP kararı: C2")
````
